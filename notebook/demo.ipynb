# AStock 演示 Notebook

本 Notebook 演示如何使用 AStock 进行盘前选股与情绪分析。

```python
import sys
sys.path.insert(0, '..')

from astock.data.akshare_data import AKShareData
from astock.strategy.zt_strategies import *
from astock.signal.engine import SignalEngine
from astock.sentiment.market import SentimentAnalyzer
from astock.execution.portfolio import Portfolio

data = AKShareData()
```

## 1. 市场情绪

```python
analyzer = SentimentAnalyzer(data)
report = analyzer.analyze()
print(analyzer.format_report(report))
```

## 2. 涨停池

```python
zt = data.get_zt_pool()
zt.head(10)
```

## 3. 策略选股

```python
engine = SignalEngine(data=data, strategies=[
    FirstBoardStrategy(),
    ContinuousBoardStrategy(),
    ResealStrategy(),
    AuctionStrategy(),
])
plans = engine.generate_plans()
print(f"共 {len(plans)} 条信号")
for p in plans[:5]:
    print(f"{p.name}({p.symbol}) | {p.strategy} | 置信度{p.confidence:.0%}")
```

## 4. 持仓查看

```python
pf = Portfolio(data)
pf.print_summary()
```
